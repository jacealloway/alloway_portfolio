```python
# Import packages 
import numpy as np 
import pandas as pd
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt 
```

```python
# Generate dataset for regression
data = make_regression(n_samples = 200, n_features=1, n_targets=1, noise = 20)

samples = data[0]
target = data[1]


plt.plot(samples, target, ls = 'none', marker = 'o')
```

```python
def linearModel(x, theta1,  theta0 = 0):
    return theta0 + x*theta1



start_point = (0, 10)
num_iterations = 50
eta = 0.1
m = 0

p_current = start_point 
while m < num_iterations:
    loss = 0

    # Calculate the inner sum for the cost function derivative for every training example 
    for i in range(len(samples)):
        loss += (1/len(samples))*(linearModel(samples[i], p_current[1]) - target[i])*samples[i]


    p_next = np.array(p_current) - np.array([0, eta * loss], dtype = object)

    p_current = p_next 
    m += 1

plt.plot(samples, target, ls = 'none', marker = 'o')
plt.plot(samples, linearModel(samples, p_current[1]))
plt.grid(alpha = 0.6)
print(p_current[1])
```

```python
# THETA_array = (X.T [m.multiply] X)^{-1} [m.multiply] (X.T [m.multiply] Y)
theta = np.dot(np.linalg.inv(np.dot(samples.T, samples)), np.dot(samples.T, target))
print(theta)
plt.plot(samples, target, ls = 'none', marker = 'o')
plt.plot(samples, samples*theta[0])

def getWeights(samples: list[float], target: list[float]) -> list[float]:
    return np.dot(np.linalg.inv(np.dot(samples.T, samples)), np.dot(samples.T, target))
```

```python
weights = LinearRegression().fit(data[0], data[1])
print(weights.coef_)

plt.plot(samples, target, ls = 'none', marker = 'o')
plt.plot(data[0].T[0], np.dot(weights.coef_[0], data[0].T[0]))
```
