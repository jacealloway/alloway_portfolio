```python
import pandas as pd 
import numpy as np 
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt 
```

```python
# Generate binary classification dataset 
data = make_classification(n_samples = 20, n_features = 2, n_informative=2, n_redundant=0, n_repeated=0, n_classes = 2)
x = data[0].T[0]
y = data[1]

plt.plot(x, y, marker = 'o', ls = 'none')
plt.show()
```

```python
# Generate logistic regression model 
def sigmoidModel(x: float, theta: float, u: float) -> float:
    return 1 / (1 + np.exp(-x*theta - u))

def lossGradientDescent(modelFunc: callable, axis, start_point: list[float], start_int: float, num_iterations: int, eta: float, samples: list[list[float]], targets: list[float]) -> list[float]:
    n_features = len(start_point)
    n_samples = len(samples)
    p_current = start_point 
    u_current = start_int

    for i in range(n_features):
        m = 0
        while m < num_iterations:
            error = 0
            int_error = 0
            for j in range(n_samples):
                error += (1/n_samples) * (targets[j] - modelFunc(samples[i][j], p_current[i], u_current)) * samples[i][j]
                int_error += (1/n_samples) * (targets[j] - modelFunc(samples[i][j], p_current[i], u_current))

            plt.plot(axis, sigmoidModel(axis, p_current[i], u_current), alpha = 0.1, color = 'red')

            p_current[i] += (eta * error) 
            u_current += (eta * int_error)
            m += 1

            
    return p_current, u_current

axis = np.linspace(np.min(x), np.max(x), 1000)
theta_mins, u_min = lossGradientDescent(sigmoidModel, axis, start_point=[1], start_int = 1, num_iterations=100, eta = 0.1, samples = [x], targets = y)

print(theta_mins[0], u_min)

plt.plot(axis, sigmoidModel(axis, theta_mins[0], u_min))
plt.plot(x, y, marker = 'o', ls = 'none')
plt.show()
```

```python
skLogRes = LogisticRegression(max_iter = 10000, random_state=0, solver = 'lbfgs').fit(x.reshape(-1, 1), y)

print(skLogRes.coef_[0][0], skLogRes.intercept_[0])
print(theta_mins[0], u_min)

plt.plot(x, y, marker = 'o', ls = 'none')
plt.plot(axis, sigmoidModel(axis, skLogRes.coef_[0][0], skLogRes.intercept_[0]), label = 'sklearn')
plt.plot(axis, sigmoidModel(axis, theta_mins[0], u_min), label = 'jace')
plt.legend(loc = 'best')
plt.show()

## Note that these values will be different because SKLearn uses different solvers. 
## Furthermore, SKLearn implements regression conditions to prevent from overfitting. The manual gradient descent does not. 
```
